```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "Qwen/Qwen3-Embedding-0.6B"
)

texts = [
    "OASIS mahsulotining narxi 150000 so'm",
    "OASIS kompaniyasining ish vaqti 09:00 dan 18:00 gacha"
]

embeddings = model.encode(texts)

print(embeddings.shape)
```

```python
# Birinchi  gap va uning 10 ta raqami
print(texts[0])
print(embeddings[0][:10])


```

```python
# Xar bir gapda qancha raqam bor 
print(len(embeddings[0]))
```

```python

# 2 ta gaplar o'xshash 
print(model.similarity(embeddings, embeddings))
```

```python
class Student:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender=gender

student1=Student('Bekzod',20,'M')
student2=Student('Nina',24,'F')
print(student1.name)
print(student2.name)
print(student1.age+student2.age)
print(student1.gender)
```
